```python
from tensorflow.keras.layers import LSTM,Dense, Embedding
from tensorflow.keras.datasets import imdb
from tensorflow.keras.models import Sequential
from tensorflow.keras.preprocessing import sequence
import numpy as np
```

```python
vocab_size=5000
(x_train,y_train),(x_test,y_test)=imdb.load_data(
    num_words=vocab_size
)
print("Training samples:", len(x_train))
print("Testing samples:", len(x_test))
```

```python
print("\nfirst review as integers:")
print(x_train[0])
word_idx = imdb.get_word_index()

# reverse the dictonary
word_idx ={i: word for word, i in word_idx.items()}
print("\nFirst review as words: ")
print([word_idx[i] for i in x_train[0] if i in word_idx])

```

```python
print(
    "\nMaximum review length: ",
    len(max((x_train + x_test),key=len))
)
print(
    "Minimum review length:",
    len(min((x_train + x_test),key=len))
)
```

```python
max_words=400
x_train= sequence.pad_sequences(
    x_train, maxlen=max_words

)
x_test = sequence.pad_sequences(
    x_test,
    maxlen=max_words
)

print("\nShape of x_train:", x_train.shape)
print("Shape of x_test: ", x_test.shape)
```

```python
x_valid=x_train[:64]
y_valid=y_train[:64]

x_train=x_train[64:]
y_train=y_train[64:]
print("\n Training data after validation split")
print(x_train.shape)
print("validation data")
print(x_valid.shape)
```

```python
embd_len=32
LSTM_model = Sequential(name="LSTM")

#convert word ids into dense vectors
LSTM_model.add(
    Embedding(
        input_dim=vocab_size,
        output_dim =embd_len
    )
)
LSTM_model.add(
    LSTM(
        128,
        activation='tanh',
        return_sequences=False
    )
)

LSTM_model.add(
    Dense(
        1,
        activation='sigmoid'
    )
)
```

```python
LSTM_model.compile(
    loss="binary_crossentropy",
    optimizer="adam",
    metrics=["accuracy"]
)
```

```python
history=LSTM_model.fit(
    x_train,
    y_train,
    batch_size=64,
    epochs=5,
    verbose=1,
    validation_data=(x_valid,y_valid)

)
```

```python
print()
test_loss, test_accuracy=LSTM_model.evaluate(
    x_test,y_test,verbose=0
)
print("LSTM Test loss:", test_loss)
print("LSTM Test accuracy:", test_accuracy)
```
