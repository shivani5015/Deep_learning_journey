"""
NER= Named Entity Recognition (NER) is an NLP technique used to find important
names and information in text such as people, place, organization ,dates and amounts and label them with their correct type"""

```python
import spacy
nlp=spacy.load("en_core_web_sm")
```

```python
#Ex 1
text="Barack Obama was born in Hawaii ana become President of thev United States"
doc=nlp(text)
for ent in doc.ents:
  print(ent.text, "->", ent.label_)
```

```python
#EX-2
text="""
Google CEO Sundar Pichai announced a $10 investment in India on 15th July 2020 """
doc=nlp(text)
for ent in doc.ents:
  print(ent.text,"->", ent.label_)
```
