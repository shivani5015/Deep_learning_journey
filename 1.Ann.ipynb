# ANN (Artificial Neural Network)
## 1. Concept
* DL model built for **tabular / structured data** (CSV files).
* Flow `Input Layer` ➔ `Hidden Layers` ➔ `Output Layer`.

## 2.Execution Loop
1. **Forward Pass (Predict):** Multiplies inputs & weights, adds bias, and runs activation to predict.
2. **Loss Check (Compare):** Calculates error (actual vs predicted).
   * Classification ➔ Cross-Entropy
   * Regression ➔ MAE / MSE
3. **Backprop(Learn and update):** Passes error backwards to adjust weights using an optimizer (`Adam` / `SGD`).

#  Activation Functions
* **ReLU:** Hidden layers default ($[0, \infty)$).
* **Leaky ReLU:** Fixes dying ReLU problem.($[-\infty, \infty)$).
* **PReLU:** Trainable negative slope parameter.($[-\infty, \infty)$).
* **Sigmoid:** Binary classification output ($(0, 1)$).
* **Softmax:** Multi-class classification output ($(0, 1)$).
* **Tanh:** Zero-centered hidden layers ($(-1, 1)$).

# Optimizers

Optimizers is used to update weight and parameter of neural network to minimise loss and improve accuracy.

* **SGD (Stochastic Gradient Descent):** it updates weights using a mini batch; simple but noisy ,slow to converge.
* **Momentum:** improves SGD by using previous weight to improves accuracy.
* **NAG (Nesterov Accelerated Gradient):** it first looks ahead at future position before taking a step, preventing overshooting.
* **AdaGrad:** it lower learning rates for frequent features while giving larger updates to rare features good for sparse data.
* **RMSprop:** Fixes AdaGrad's learning rate problem using an exponential moving average.
* **Adam (Adaptive Moment Estimation):** Combines Momentum + RMSprop; default go-to optimizer for almost every deep learning model.

```python
import pandas as pd
import numpy as np
```

```python
df=pd.read_csv("/content/Attrition - Attrition.csv")  #load data
df.head()
```

```python
import tensorflow as tf       #import libraries
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder,StandardScaler
from sklearn.metrics import accuracy_score
```

```python
df=df.dropna()
```

```python
x=df.drop(columns=['Attrition'],axis=1)  #divide data
y=df['Attrition']
```

```python
categorical_features=['BusinessTravel','Department','EducationField','Gender','JobRole','MaritalStatus','Over18','OverTime']
```

```python
x_encode=pd.get_dummies(x,columns=categorical_features,drop_first=True)
```

```python
x_encode=x_encode.astype(int)
```

```python
lb=LabelEncoder()
y_encode=lb.fit_transform(y)
y_encode
```

```python
x_train,x_test,y_train,y_test=train_test_split(x_encode,y_encode,test_size=0.2,random_state=42)
```

```python
x_train.head()
```

```python
sc=StandardScaler()
x_train_sc=sc.fit_transform(x_train)
x_test_sc=sc.transform(x_test)
```

```python
model=tf.keras.models.Sequential([
    tf.keras.layers.Dense(units=128,activation='relu',input_dim=x_train_sc.shape[1]),
    tf.keras.layers.Dropout(0.3),
    tf.keras.layers.Dense(units=64,activation='relu'),
    tf.keras.layers.Dropout(0.3),
    tf.keras.layers.Dense(units=1,activation='sigmoid')
])
```

```python
loss_fn = tf.keras.losses.BinaryCrossentropy()
optimizer = tf.keras.optimizers.SGD(learning_rate=0.001, momentum=0.9, nesterov=True)

model.compile(optimizer=optimizer, loss=loss_fn, metrics=['accuracy'])
```

```python
model.fit(x_train_sc, y_train, epochs=50, batch_size=8, validation_split=0.1)
```

```python
loss, accuracy = model.evaluate(x_test_sc, y_test)
print(f"test_loss={loss:.4f}, test_accuracy={accuracy:.4f}")
```
