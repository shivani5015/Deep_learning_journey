```python
from gensim.models import Word2Vec
from gensim.utils import simple_preprocess


raw_corpus = [
    "Data Science uses Machine Learning",
    "Gensim library is quite useful for NLP tasks",
    "Word2Vec model gives us word embeddings",
    "Deep Learning and Neural Networks improve NLP",
    "FastText uses subword information for embeddings"
]

#tokensization
tokensized_corpus=[simple_preprocess(sentence) for sentence in raw_corpus]
print(tokensized_corpus)
```

```python
#step 2 : Train the Word2Vec Model
```

```python
model= Word2Vec(
    sentences=tokensized_corpus,
    vector_size=50,  #dimension of embedding
    window=3,        #context window size
    min_count=1,     #minimum word frequency
    sg=1,            #1=skip-gram,0=CBOW
    workers=4        #parallel threads(use the multiple  core of cpu  for faster training)
)
```

```python
#step 3= use the model
```

```python
#show the vector of any word
print(model.wv["data"])
print(model.wv["data"].shape)
```

```python
#see the similar words
print(model.wv.most_similar("science",topn=3))
```

```python
#check the similiarity of 2 words
print(model.wv.similarity("data","science"))
```
